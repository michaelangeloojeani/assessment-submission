### 1. 32-bit Words

A 32-bit word can be represented using a Python integer (int).

Since Python integers can exceed 32 bits, 0xFFFFFFFF is used to keep the value within 32 bits.



```python

# Import NumPy.
import numpy as np

# Create a 32-bit unsigned integer.
word = np.uint32(0x12345678)

# Display the word in hexadecimal.
print(f"{word:08x}")

# Display its size in bits.
print(word.dtype.itemsize * 8)
```

### 2. Sequences of 32-bit Words

Python lists (list) can store multiple 32-bit words.

```python
# Created an array of 32-bit words.
words = np.array([0x12345678, 0xABCDEF12, 0x98765432], dtype=np.uint32)

# Display each word in hexadecimal.
for word in words:
    print(f"{word:08x}")
```

### 3. Input Messages


```python
# Create a text message.
message = "Hello World"

# Convert the message into bytes.
message_bytes = message.encode("utf-8")

# Display the bytes.
print(message_bytes)
```

### 4. 512-bit Message Blocks

```python
# Create a 512-bit block containing 16 words.
block = np.zeros(16, dtype=np.uint32)

# Display the block.
print(block)

# Display the total size in bits.
print(block.nbytes * 8)
```

### 5. Final 256-bit Hash Value

```python
# Create eight 32-bit words.
hash_words = np.zeros(8, dtype=np.uint32)

# Display the words in hexadecimal.
hash_hex = "".join(f"{word:08x}" for word in hash_words)

print(hash_hex)
print(len(hash_hex) * 4)
```

### 6. Big-Endian Integers
Big-endian stores the most significant byte first.

I chose "int.from_bytes()" because it converts bytes into an integer using the byte order required by SHA-256.

```python
# Create four bytes.
data = bytes([0x12, 0x34, 0x56, 0x78])

# Convert the bytes into a big-endian integer.
number = int.from_bytes(data, byteorder="big")

# Display the result in hexadecimal.
print(f"{number:08x}")
```
