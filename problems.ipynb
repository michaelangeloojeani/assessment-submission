### 1. 32-bit Words

A 32-bit word is an integer containing exactly 32 bits. SHA-256 uses these words during its calculations and processing.

Python integers can represent values larger than 32 bits. Therefore, a bitwise AND operation with `0xFFFFFFFF` can be used to ensure that a value stays within the 32-bit range.

A 32-bit word can represent values between 0 and 4,294,967,295.

```python
words = [
    0x12345678,
    0xABCDEF12,
    0x98765432
]

for word in words:
    print(hex(word))
```
